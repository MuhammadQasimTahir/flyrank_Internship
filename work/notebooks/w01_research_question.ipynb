## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The Research Question: Which content items (pages) should be reviewed first for refresh, expansion, protection, pruning, or monitoring?  

Unit of Analysis: A single content item (page).  

Decision Supported: Resource allocation for content maintenance. Rather than manually auditing thousands of pages, the team uses the ranked queue to direct human attention.  

Action Taken: An editor or SEO specialist inspects the top-ranked candidate pages alongside their reason codes to update, rewrite, consolidate, or leave the page as-is.  

Cost of a Wrong Call:False Positive (Waste): A high-ranking recommendation for a page that does not need work wastes human review time and editing budget.

False Negative (Opportunity Loss): Failing to surface a declining, high-demand page leaves decaying search visibility and traffic unaddressed.

Why Data/ML Helps: While simple heuristic rules (e.g., stale AND high_volume) flag obvious outliers, they fail to evaluate nuanced interactions across the middle inventory—such as position-tier CTR expectations versus engagement drop-offs. Machine learning models learn non-linear patterns across these observable signals to deliver a more accurate ranking.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

A preliminary audit of the starter dataset demonstrates why automated prioritization is necessary: over half of the inventory shows a downward performance trend, and a vast majority of pages have not been updated in over six months

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import sys
import subprocess
import pandas as pd

# 1. Environment & Path Setup (Handles Colab and local execution safely)
if "google.colab" in sys.modules:
    REPO_DIR = "flyrank-ml-internship-starter"
    if not os.path.isdir(REPO_DIR):
        # Clones repo into Colab if not present
        subprocess.run(["git", "clone", "--depth", "1", "https://github.com/flyrank-bih/flyrank-ml-internship-starter", REPO_DIR], check=True)
    os.chdir(REPO_DIR)
elif os.path.basename(os.getcwd()) == "notebooks":
    # Moves up to root if running from inside work/notebooks/
    os.chdir("../..")

# 2. Loading Starter Dataset
data_path = "data/raw/content_refresh_anonymized.csv"
assert os.path.exists(data_path), f"Starter data not found at {data_path}. Ensure working directory is repo root."
df = pd.read_csv(data_path)

# 3. Compute Supporting Metrics
total_pages = len(df)
declining_rate = (df["trend_direction"].str.lower() == "down").mean()
stale_rate = (df["days_since_last_update"] >= 180).mean()
visible_rate = (df["impressions_90d"] >= 500).mean()

# Output Results
print(f"--- Inventory Snapshot (N = {total_pages:,} pages) ---")
print(f"1. Declining Trend Rate:       {declining_rate:.1%} ({int(declining_rate * total_pages):,} pages)")
print(f"2. Stale Rate (>= 180 days):    {stale_rate:.1%} ({int(stale_rate * total_pages):,} pages)")
print(f"3. High-Visibility (>= 500 imp): {visible_rate:.1%} ({int(visible_rate * total_pages):,} pages)")
```
