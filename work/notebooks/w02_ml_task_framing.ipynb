## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success Metric: Precision@K (Specifically, Precision@50).  

What means 'good': Overall model accuracy is irrelevant if the top recommendations are wrong. Because editorial capacity is limited, we care exclusively about the top of the ranked list. If a reviewer can only audit 50 pages this week, Precision@50 measures exactly how many of those top 50 pages were actually declining. A "good" score means the model significantly outperforms the hand-written baseline (e.g., catching ~35+ true declines in the top 50 versus the baseline's ~12).

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why ML outperforms a fixed rule:Fixed SEO rules (like days_since_last_update > 180 AND impressions > 500) are rigid and create massive ties. Similar to evaluating complex credit risk profiles, where multiple interacting financial signals outclass flat decision thresholds, relying on strict cutoffs here fails to capture the nuance of search behavior.  

An ML algorithm (like a Random Forest or Gradient Boosting model) captures non-linear interactions. It learns how a page's age interacts differently depending on its specific position tier, CTR gap, and competition level, allowing it to break ties intelligently and cleanly stack-rank the "messy middle" of the inventory.
